```python
import numpy as np

def whoami_cpu():
    # Configuración de bloques y hilos
    b_x, b_y = 2, 2
    t_x, t_y = 4, 1

    grid_dim_x, grid_dim_y = b_x, b_y
    block_dim_x, block_dim_y = t_x, t_y

    total_threads = b_x * b_y * t_x * t_y

    for block_y in range(b_y):
        for block_x in range(b_x):
            # ID lineal del bloque (2D → 1D)
            block_id = block_x + block_y * grid_dim_x

            for thread_y in range(t_y):
                for thread_x in range(t_x):
                    # ID del hilo dentro del bloque (2D → 1D)
                    thread_offset = thread_x + thread_y * block_dim_x

                    # Offset global del bloque
                    block_offset = block_id * (block_dim_x * block_dim_y)

                    # ID global del hilo
                    global_id = block_offset + thread_offset

                    # Impresión
                    print(f"{global_id:03d} | Block ({block_x},{block_y}) = {block_id:2d} | "
                          f"Thread ({thread_x},{thread_y}) = {thread_offset:2d} | "
                          f"BlockDim ({block_dim_x},{block_dim_y}) GridDim ({grid_dim_x},{grid_dim_y})")

# Ejecutar la función
whoami_cpu()
```

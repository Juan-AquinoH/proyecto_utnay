```python
from numba import cuda

@cuda.jit
def whoami():
    # ID lineal del bloque (2D → 1D)
    block_id = cuda.blockIdx.x + cuda.blockIdx.y * cuda.gridDim.x

    # Hilos por bloque
    threads_per_block = cuda.blockDim.x * cuda.blockDim.y

    # Offset global del bloque
    block_offset = block_id * threads_per_block

    # ID del hilo dentro del bloque (2D → 1D)
    thread_offset = cuda.threadIdx.x + cuda.threadIdx.y * cuda.blockDim.x

    # ID global del hilo
    global_id = block_offset + thread_offset

    print(f"{global_id:03d} | Block ({cuda.blockIdx.x},{cuda.blockIdx.y}) = {block_id:2d} | "
          f"Thread ({cuda.threadIdx.x},{cuda.threadIdx.y}) = {thread_offset:2d} | "
          f"BlockDim ({cuda.blockDim.x},{cuda.blockDim.y}) GridDim ({cuda.gridDim.x},{cuda.gridDim.y})")


# Configuración
b_x, b_y = 2, 2
t_x, t_y = 4, 1

blocks_per_grid = (b_x, b_y)
threads_per_block = (t_x, t_y)

whoami[blocks_per_grid, threads_per_block]()
cuda.synchronize()
```
